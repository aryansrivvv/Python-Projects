```python
import numpy as np
states = ['A', 'B', 'C', 'D', 'E'] # Total states - 5 , Terminal States - A, E 
terminal_states = ['A' , 'E'] 
num_states = len(states)
rewards =  [ 0 , 0 ,0 , 0 , 1] 
```

```python
# Transition probabilities for a random walk
# Each state transitions to its adjacent states with equal probability
transition_matrix = [
     [ 0 , 1 , 0 , 0 , 0 ] , 
     [ 0.5 , 0 , 0.5 , 0 , 0 ] , 
     [ 0 , 0.5 , 0 , 0.5 , 0 ] ,
     [ 0 , 0 , 0.5 , 0 , 0.5 ] ,
     [ 0 , 0 , 0 , 1 , 0 ] ,
]
# Discount factor for TD(0) algorithm
discount = 1 # No discounting for this example
true_rewards = {'A': 0, 'E': 1}
```

```python
import random 
def choose_random_non_zero(arr):
    # Create a list of indices corresponding to non-zero elements
    non_zero_indices = [i for i, value in enumerate(arr) if value != 0]
    # Randomly choose an index from the list of non-zero indices
    random_index = random.choice(non_zero_indices)
    random_non_zero_number = arr[random_index]
    return random_non_zero_number, random_index
```

```python
def td_0_random_walk(transition_matrix, rewards, discount , num_episodes=100 , alpha=0.1):
    value_function = [1/6,0,0,0,5/6]
    for _ in range(num_episodes) :
        current_state = 2  # corresponding to 'C'
        while (True) :
            probability_of_next_state , next_state = choose_random_non_zero(transition_matrix[current_state])
            reward  = rewards[current_state] 
            value_function[current_state] += alpha * (reward + discount * value_function[next_state] - value_function[current_state])
            current_state = next_state
            if current_state == 0 or current_state == 4:
                break
    return value_function
```
