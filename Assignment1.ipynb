```python
#Assignment 1 
import numpy as np

num_arms = 10
true_action_values = np.random.normal(0, 1, num_arms)
```

```python
def epsilon_greedy(Q, epsilon):
    if np.random.random() < epsilon:
        # Explore: choose a random action
        return np.random.choice(len(Q))
    else:
        # Exploit: choose the action with the maximum estimated value
        return np.argmax(Q)

# Initialize action value estimates for epsilon-greedy
Q_epsilon_greedy = np.zeros(num_arms)
```

```python
# Initialize action value estimates with an optimistic initial value
initial_value = 5.0  # Choose a suitable optimistic initial value
Q_optimistic = np.full(num_arms, initial_value)
```

```python
def ucb(Q, N, t):
    ucb_values = Q + np.sqrt(np.log(t) / (N + 1e-6))
    return np.argmax(ucb_values)

# Initialize action value estimates and action visit counts for UCB
Q_ucb = np.zeros(num_arms)
N_ucb = np.ones(num_arms)  # Each action has been selected once initially
```

```python
# Number of time steps or iterations
num_steps = 1000

# Simulation
for step in range(num_steps):
    # Choose actions for each method
    action_epsilon_greedy = epsilon_greedy(Q_epsilon_greedy, epsilon=0.1)
    action_optimistic = np.argmax(Q_optimistic)
    action_ucb = ucb(Q_ucb, N_ucb, step + 1)

    # Simulate rewards for each action (based on true action values)
    reward_epsilon_greedy = np.random.normal(true_action_values[action_epsilon_greedy], 1)
    reward_optimistic = np.random.normal(true_action_values[action_optimistic], 1)
    reward_ucb = np.random.normal(true_action_values[action_ucb], 1)

    # Update action value estimates
    Q_epsilon_greedy[action_epsilon_greedy] += (reward_epsilon_greedy - Q_epsilon_greedy[action_epsilon_greedy]) / (step + 1)
    Q_optimistic[action_optimistic] += (reward_optimistic - Q_optimistic[action_optimistic]) / (step + 1)
    Q_ucb[action_ucb] += (reward_ucb - Q_ucb[action_ucb]) / (N_ucb[action_ucb])
    N_ucb[action_ucb] += 1

# After the simulation, Q_epsilon_greedy, Q_optimistic, and Q_ucb contain the estimated action values for each method.
```
