```python
import pandas as pd
import numpy as np
import io

print(f"Pandas Version: {pd.__version__}")
```

```python
# ==========================================
# Dataset 1: Employee HR Data
# ==========================================
data_hr = {
    'emp_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'name': ['Alice Smith', 'Bob Jones', 'Charlie Brown', 'David Lee', 'Eva Green',
             'Frank Wright', 'Grace Hopper', 'Hannah Abbott', 'Ian Malcolm', 'Julia Roberts'],
    'department': ['Sales', 'Engineering', 'Engineering', 'HR', 'Sales',
                   'Engineering', 'HR', 'Sales', 'Marketing', np.nan],
    'age': [28, 34, 22, 45, 29, 52, 38, np.nan, 41, 31],
    'salary': [65000, 95000, 70000, 58000, 68000, 120000, 62000, 54000, 80000, 72000],
    'join_date': ['2020-01-15', '2018-03-22', '2021-07-01', '2015-11-10', '2019-05-19',
                  '2010-02-01', '2017-08-14', '2022-04-05', '2016-09-30', '2020-12-01'],
    'performance_score': [4.2, 4.8, 3.5, 3.9, 4.1, 4.9, 3.8, np.nan, 4.5, 4.0]
}

df_hr = pd.DataFrame(data_hr)
df_hr['join_date'] = pd.to_datetime(df_hr['join_date'])
```

```python
# ==========================================
# Dataset 2: E-Commerce Sales Transactions
# ==========================================
data_sales = {
    'order_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'emp_id': [101, 101, 102, 105, 108, 103, 101, 109],
    'product': ['Laptop', 'Mouse', 'Monitor', 'Keyboard', 'Mouse', 'Laptop', 'Monitor', 'Headphones'],
    'quantity': [1, 2, 1, 1, 3, 2, 1, 2],
    'unit_price': [1200, 25, 300, 75, 25, 1150, 310, 150]
}

df_sales = pd.DataFrame(data_sales)

print("✅ Data successfully loaded!")
print("\n--- HR DataFrame Preview ---")
display(df_hr.head())
```

```python
print("--- 1. Dimensions & Structure ---")
print("Shape (Rows, Columns):", df_hr.shape)
print("Column Names:", df_hr.columns.tolist())
print("\n--- 2. Data Types & Memory Usage ---")
df_hr.info()

print("\n--- 3. Summary Statistics for Numeric Columns ---")
display(df_hr.describe())

print("\n--- 4. Summary Statistics for Categorical/All Columns ---")
display(df_hr.describe(include='all'))

print("\n--- 5. Count Unique Values ---")
print("Unique departments:", df_hr['department'].nunique())
print("\nValue Counts (including NaNs):")
print(df_hr['department'].value_counts(dropna=False))
```

```python
# 1. Column Selection
dept_series = df_hr['department']             # Single column -> Series
sub_df = df_hr[['name', 'salary', 'age']]     # Multiple columns -> DataFrame

# 2. .loc[] (Label-based indexing: loc[row_labels, column_names])
print("--- .loc Example (Rows 0 to 2, specific columns) ---")
display(df_hr.loc[0:2, ['name', 'department', 'salary']])

# 3. .iloc[] (Position-based indexing: iloc[row_indices, col_indices])
print("\n--- .iloc Example (First 3 rows, first 3 columns) ---")
display(df_hr.iloc[0:3, 0:3])

# 4. Conditional Filtering (Use & for AND, | for OR, ~ for NOT)
print("\n--- Filter: Engineering Dept AND Salary > $80,000 ---")
eng_high = df_hr[(df_hr['department'] == 'Engineering') & (df_hr['salary'] > 80000)]
display(eng_high[['name', 'department', 'salary']])

# 5. .isin() and .between()
print("\n--- Filter: Department in ['Sales', 'HR'] ---")
display(df_hr[df_hr['department'].isin(['Sales', 'HR'])][['name', 'department']])

print("\n--- Filter: Age between 25 and 35 ---")
display(df_hr[df_hr['age'].between(25, 35)][['name', 'age']])

# 6. .query() method (Clean SQL-like expression syntax)
print("\n--- .query() Example: Salary > 70000 and Age < 40 ---")
display(df_hr.query("salary > 70000 and age < 40")[['name', 'salary', 'age']])
```

```python
df_clean = df_hr.copy()

print("--- Missing Values Count per Column ---")
print(df_clean.isna().sum())

# 1. Fill Missing Values
df_clean['department'] = df_clean['department'].fillna('Unassigned') # Fill constant
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())    # Impute median
df_clean['performance_score'] = df_clean['performance_score'].ffill() # Forward-fill

# 2. Drop Missing Values (Alternative)
# df_clean.dropna(subset=['salary'], inplace=True)

# 3. Handle Duplicates
df_dup = pd.concat([df_clean, df_clean.iloc[[0, 1]]])  # Simulate duplicate rows
print(f"\nDuplicates count before dropping: {df_dup.duplicated().sum()}")

df_dedup = df_dup.drop_duplicates(subset=['emp_id'], keep='first')
print(f"Shape after dropping duplicates: {df_dedup.shape}")

# 4. String Cleaning Operations (.str accessor)
df_clean['name_upper'] = df_clean['name'].str.upper()
df_clean['first_name'] = df_clean['name'].str.split(' ').str[0]
df_clean['has_smith'] = df_clean['name'].str.contains('Smith', case=False)

display(df_clean[['name', 'first_name', 'department', 'age', 'performance_score']].head())
```

```python
# 1. Simple Calculations
df_hr['bonus'] = df_hr['salary'] * 0.10

# 2. Binary Logic with np.where
df_hr['is_senior'] = np.where(df_hr['age'] >= 40, 'Senior', 'Junior')

# 3. Multi-condition Logic with np.select
conditions = [
    df_hr['salary'] < 60000,
    (df_hr['salary'] >= 60000) & (df_hr['salary'] < 90000),
    df_hr['salary'] >= 90000
]
choices = ['Entry Level', 'Mid Level', 'Executive']
df_hr['salary_tier'] = np.select(conditions, choices, default='Unknown')

# 4. Binning Continuous Data
# pd.cut: Equal width bins / custom edges
df_hr['age_group'] = pd.cut(df_hr['age'], bins=[20, 30, 40, 60], labels=['20s', '30s', '40+'])

# pd.qcut: Equal size / Quantile-based bins
df_hr['salary_quantile'] = pd.qcut(df_hr['salary'], q=3, labels=['Low', 'Medium', 'High'])

# 5. Applying Custom Functions (.apply)
def grade_performance(score):
    if pd.isna(score): return 'N/A'
    return 'Exceeds Expectations' if score >= 4.5 else 'Meets Expectations'

df_hr['perf_grade'] = df_hr['performance_score'].apply(grade_performance)

display(df_hr[['name', 'salary', 'salary_tier', 'age_group', 'salary_quantile', 'perf_grade']])
```

```python
print("--- 1. Basic GroupBy Aggregation ---")
dept_avg = df_hr.groupby('department')['salary'].mean()
print(dept_avg)

print("\n--- 2. Advanced Multi-Metric Aggregation (.agg) ---")
dept_summary = df_hr.groupby('department').agg(
    employee_count=('emp_id', 'count'),
    avg_salary=('salary', 'mean'),
    max_salary=('salary', 'max'),
    avg_perf=('performance_score', 'mean')
).reset_index()

display(dept_summary)

print("\n--- 3. Pivot Table ---")
pivot_tbl = df_hr.pivot_table(
    values='salary',
    index='department',
    columns='salary_tier',
    aggfunc='count',
    fill_value=0
)
display(pivot_tbl)

print("\n--- 4. Crosstab (Frequency matrix) ---")
ct = pd.crosstab(df_hr['department'], df_hr['is_senior'], margins=True)
display(ct)
```

```python
print("--- 1. SQL-Style Merge (Join) ---")
# Merge Sales transactions with HR employee details on 'emp_id'
df_merged = pd.merge(df_sales, df_hr, on='emp_id', how='left')

df_merged['total_revenue'] = df_merged['quantity'] * df_merged['unit_price']

display(df_merged[['order_id', 'name', 'department', 'product', 'quantity', 'total_revenue']])

print("\n--- 2. Concatenating DataFrames (Stacking Rows/Columns) ---")
part1 = df_hr.iloc[:3]
part2 = df_hr.iloc[3:6]

# Stack vertically
recombined_df = pd.concat([part1, part2], axis=0, ignore_index=True)
print(f"Recombined DataFrame rows: {len(recombined_df)}")
```

```python
# Create a 30-day Time Series dataset
dates = pd.date_range(start='2026-01-01', periods=30, freq='D')
np.random.seed(42)
daily_sales = np.random.randint(100, 500, size=30)

df_ts = pd.DataFrame({'date': dates, 'sales': daily_sales})
df_ts.set_index('date', inplace=True)

# 1. DateTime Attribute Extraction
df_ts['day_name'] = df_ts.index.day_name()
df_ts['is_weekend'] = df_ts.index.dayofweek >= 5

# 2. Resampling (e.g. Daily to Weekly)
weekly_sales = df_ts['sales'].resample('W').sum()
print("--- Weekly Aggregated Sales ---")
print(weekly_sales)

# 3. Rolling Windows & Lags
df_ts['7_day_MA'] = df_ts['sales'].rolling(window=7).mean()  # Moving Average
df_ts['prev_day_sales'] = df_ts['sales'].shift(1)            # Lag column
df_ts['pct_change'] = df_ts['sales'].pct_change() * 100       # % Growth

print("\n--- Time Series Calculations (First 10 Days) ---")
display(df_ts.head(10))
```
