```python
# Reverse the array
def reverse_array(arr):
  l = len(arr)
  for i in range(l//2):
    arr[i],arr[l-i-1] = arr[l-i-1] , arr[i]
  return arr

arr = [1, 4, 3, 2, 6,5]
print('Approach 1- ' , reverse_array(arr))

# two pointer approach

def reverse_arr_two_ptr(arr):
  p = 0
  q = len(arr)-1
  while(p<q):
    arr[p],arr[q]= arr[q],arr[p]
    p+=1
    q-=1
  return arr

print('Two pointer approach -' , reverse_arr_two_ptr(arr))
```

```python
# find max and min in an array

arr = [1, -44, 3, 200, 6,0]
def find_max_min(arr): # O(n) time O(log n) space , 2n comparisions
  minn = float('inf')
  maxx = float('-inf')
  for i in arr:
    minn = min(minn,i)
    maxx = max(maxx,i)
  return (minn,maxx)

print('approach1-(min,max)' , find_max_min(arr))

def tournament_method(arr,low  , high): # O(n) time O(log n) space but lesser comparisions (3n/2)
  if ( low == high ): # only one element
    return (arr[low] , arr[high])
  elif (low == high - 1): # two consecutive elements
    if ( arr[low] < arr[high]):
      return (arr[low],arr[high])
    else :
      return (arr[high],arr[low])
  else:  # dividing the array and recursively calling
    mid = (low + high)//2
    left_result = tournament_method(arr,low,mid)
    right_result = tournament_method(arr,mid+1,high )
    final_min = min(left_result[0], right_result[0])
    final_max = max(left_result[1],right_result[1])
    return (final_min,final_max)

data = [1000, 11, 445, 1, 330, 3000]
l = len(data)
print('Divide and conquer-', tournament_method(data,0,l-1))

def optimal_approach(arr): # 3n/2 comparisions with O(1) space
  l = len(arr)
  if(l==0): return None
  if(l==1): return (arr[0],arr[1]) # comparing first two pairs manually and then comparing two pairs together reducing comparisions by half
  if (l%2==0):
    if (arr[0]< arr[1]):
      minn = arr[0]
      maxx = arr[1]
    else:
      minn = arr[1]
      maxx = arr[0]
    i = 2
  else:
    minn = maxx = arr[0]
    i = 1
  while(i<l-1):
    if(arr[i]< arr[i+1]):
      minn = min(arr[i],minn)
      maxx = max(arr[i+1],maxx)
    else:
      minn = min(arr[i+1],minn)
      maxx = max(arr[i],maxx)
    i+=2
  return (minn,maxx)

data_1 = [1000, 11, 445, 1, 330, 3000, 600, -12001]
print('Optimal Approach-', optimal_approach(data_1))
```
